🧾 **Money Wiz – Customer and Sales Management System**

🎯 **Objective**
This Python project simulates a basic Customer Relationship Management (CRM) and Sales Visualization System.
It allows users (Salesperson or Manager) to perform customer data operations and visualize business insights using Excel files.

1. **File Handling Functions**

There are two helper functions for reading and writing Excel files.

**read_excel()** → Reads an Excel sheet into a DataFrame.
**write_excel()** → Writes updated data back to Excel.

```python
import pandas as pd
import matplotlib.pyplot as plt

data_FILE = "customers.xlsx"

# file reading function
def read_excel(filename, sheet_name=0):
    return pd.read_excel(filename, sheet_name=sheet_name)

# file writing function
def write_excel(filename, df, sheet_name="Sheet1"):
       with pd.ExcelWriter(filename) as writer:
           df.to_excel(writer, index=False, sheet_name=sheet_name)
```

2. **Customer Management Functions**

**Purpose:**

Allow salespersons to perform CRUD operations (Create, Read, Update, Delete) for customer operations and personal sales visualization.

**Operations:**

**find_customer_by_name()** – Searches customers by name, used as common helper function.
**add_customer()** – Adds new customer details to the Excel file.
**update_customer()** – Lets users modify existing records interactively.
**delete_customer()** – Removes a customer record by name.
**view_customer()** – Displays full details of a selected customer.
**past_product_sales()** - Displays visual details for total amount of past sales of products for specifc salesperson.

```python
# Customer Functions

#common function to find customer in dataset
def find_customer_by_name(df, name):
    result = df[df["CustomerName"] == name]
    return result if not result.empty else None

#customer record addition function
def add_customer():
    df = read_excel(data_FILE)

    # getting customer data inputs
    CustomerName = input("customerName: ")
    CustomerType = input("customerType: ")
    Product = input("product: ")
    Quantity = int(input("quantity: "))
    Region = input("region: ")
    Date = input("date: ")
    UnitPrice = float(input("UnitPrice: "))
    StoreLocation = input("StoreLocation: ")
    Discount = float(input("Discount: "))
    Salesperson = input("Salesperson: ")
    TotalPrice = float(input("TotalPrice: "))
    PaymentMethod = input("PaymentMethod: ")
    Promotion = input("Promotion: ")
    Returned = input("Returned: ")
    OrderID = input("OrderID: ")
    ShippingCost = float(input("ShippingCost: "))
    OrderDate = input("OrderDate: ")
    DeliveryDate = input("DeliveryDate: ")
    RegionManager = input("RegionManager: ")

    # add records in columns
    new_row = pd.DataFrame({
        'CustomerName': [CustomerName],
        'CustomerType': [CustomerType],
        'Product': [Product],
        'Quantity': [Quantity],
        'Region': [Region],
        'Date': [Date],
        'UnitPrice': [UnitPrice],
        'StoreLocation': [StoreLocation],
        'Discount': [Discount],
        'Salesperson': [Salesperson],
        'TotalPrice': [TotalPrice],
        'PaymentMethod': [PaymentMethod],
        'Promotion': [Promotion],
        'Returned': [Returned],
        'OrderID': [OrderID],
        'ShippingCost': [ShippingCost],
        'OrderDate': [OrderDate],
        'DeliveryDate': [DeliveryDate],
        'RegionManager': [RegionManager]
        })
    df = pd.concat([df, new_row], ignore_index=True)
    #used concat as append function as getting function deprecated error
    write_excel(data_FILE, df)
    print("Customer added successfully.")

#customer record updation fumction
def update_customer():
    df = read_excel(data_FILE)

    #getting name of customer whose data is to be updated
    name = input("Enter customer name to update: ")
    existing = find_customer_by_name(df, name)

    #checking if customer data exist in dataset
    if existing is None:
        print("Customer not found!")
        return

    #getting data for fields to be updated
    customerType_input = input("New customerType: ")
    if customerType_input != "":
        df.loc[df["CustomerName"] == name, "CustomerType"] = customerType_input

    product_input = input("New product: ")
    if product_input != "":
        df.loc[df["CustomerName"] == name, "Product"] = product_input

    quantity_input = input("New quantity: ")
    if quantity_input != "":
        df.loc[df["CustomerName"] == name, "Quantity"] = float(quantity_input)

    region_input = input("New region: ")
    if region_input != "":
        df.loc[df["CustomerName"] == name, "Region"] = region_input

    date_input = input("New date: ")
    if date_input != "":
        df.loc[df["CustomerName"] == name, "Date"] = date_input

    UnitPrice_input = input("New UnitPrice: ")
    if UnitPrice_input != "":
        df.loc[df["CustomerName"] == name, "UnitPrice"] = float(UnitPrice_input)

    StoreLocation_input = input("New StoreLocation: ")
    if StoreLocation_input != "":
        df.loc[df["CustomerName"] == name, "StoreLocation"] = StoreLocation_input

    Discount_input = input("New Discount: ")
    if Discount_input != "":
        df.loc[df["CustomerName"] == name, "Discount"] = float(Discount_input)

    Salesperson_input = input("New Salesperson: ")
    if Salesperson_input != "":
        df.loc[df["CustomerName"] == name, "Salesperson"] = Salesperson_input

    TotalPrice_input = input("New TotalPrice: ")
    if TotalPrice_input != "":
        df.loc[df["CustomerName"] == name, "TotalPrice"] = float(TotalPrice_input)

    PaymentMethod_input = input("New PaymentMethod: ")
    if PaymentMethod_input != "":
        df.loc[df["CustomerName"] == name, "PaymentMethod"] = PaymentMethod_input

    Promotion_input = input("New Promotion: ")
    if Promotion_input != "":
        df.loc[df["CustomerName"] == name, "Promotion"] = Promotion_input

    Returned_input = input("New Returned: ")
    if Returned_input != "":
        df.loc[df["CustomerName"] == name, "Returned"] = Returned_input

    OrderID_input = input("New OrderID: ")
    if OrderID_input != "":
        df.loc[df["CustomerName"] == name, "OrderID"] = OrderID_input

    ShippingCost_input = input("New ShippingCost: ")
    if ShippingCost_input != "":
        df.loc[df["CustomerName"] == name, "ShippingCost"] = float(ShippingCost_input)

    OrderDate_input = input("New OrderDate: ")
    if OrderDate_input != "":
        df.loc[df["CustomerName"] == name, "OrderDate"] = OrderDate_input

    DeliveryDate_input = input("New DeliveryDate: ")
    if DeliveryDate_input != "":
        df.loc[df["CustomerName"] == name, "DeliveryDate"] = DeliveryDate_input

    RegionManager_input = input("New RegionManager: ")
    if RegionManager_input != "":
        df.loc[df["CustomerName"] == name, "RegionManager"] = RegionManager_input

    write_excel(data_FILE, df)
    print("Customer updated successfully.")

#customer record deletion fumction
def delete_customer():
    df = read_excel(data_FILE)

    #getting name of customer whose data is to be deleted
    name = input("Enter customer name to delete: ")
    existing = find_customer_by_name(df, name)

    #checking if customer data exist in dataset 
    if existing is None:
        print("Customer not found!")
        return

    df = df[df["CustomerName"] != name]
    write_excel(data_FILE, df)
    print("Customer deleted successfully.")

#customer record viewing fumction
def view_customer():
    df = read_excel(data_FILE)

    #getting name of customer whose data is to be displayed
    name = input("Enter customer name to view: ")
    existing = find_customer_by_name(df, name)

    #checking if customer data exist in dataset
    if existing is None:
        print("Customer not found!")
    else:
        print("Customer details:")
        print(existing)

#Display and plot past sales (TotalPrice) per Product for a given salesperson.
def past_product_sales():
    
    salesperson_name = input("Enter salesperson name: ")
    # Read data and filter for the salesperson
    df = read_excel(data_FILE)
    salesperson_df = df[df["Salesperson"].str.strip().str.lower() == salesperson_name.strip().lower()]
    
    if salesperson_df.empty:
        print(f"No sales records found for salesperson: {salesperson_name}")
        return
    
    # Group by product and sum total prices
    product_sales = salesperson_df.groupby("Product")["TotalPrice"].sum()
    
    # Print summary
    print(f"\nSales Summary for {salesperson_name}:")
    for product, total in product_sales.items():
        print(f"  {product}: €{total:.2f}")
    
    # Plot results
    product_sales.plot(kind='bar', color='skyblue', edgecolor='black', figsize=(8, 4))
    plt.title(f"Total Sales per Product - {salesperson_name}")
    plt.xlabel("Product")
    plt.ylabel("Total Sales (€)")
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()
```

3. **Region Manager Operations & Visualization**

**Purpose:**

To share visualization of sales data for the region managers for analysis

**Operations:**

**region_wise_sale()** - Shows total sales per product in a region
**store_wise_sale()** - Shows total sales per product in a specific store
**salesperson_wise_sale()** - Shows product-wise sales by a salesperson
**max_product_sales_by_store()** - Identifies the best-selling product for each store
**salesperson_max_sales()** - Highlights the top-performing salesperson
**store_wise_return()** - Shows total returned items per product in a store

```python
# Manager Operations

#Region specifc sales graph
def region_wise_sale():
    df = read_excel(data_FILE)
    region = input("Enter region name: ")
    # Filter rows where 'Region' column from dataset matches the given region name.
    result = df[df["Region"].astype(str) == region]

    # Group by product and sum total prices
    summary = result.groupby("Product", as_index=False)["TotalPrice"].sum()

    # Plot results
    plt.figure(figsize=(10, 5))
    bars = plt.bar(summary["Product"].astype(str), summary["TotalPrice"])
    plt.xlabel("Product")
    plt.ylabel("Total Price")
    plt.title(f"Sales in Region: {region}")
    plt.xticks(rotation=45)
    plt.bar_label(bars, fmt='%.2f', fontsize=9)
    plt.tight_layout()
    plt.show()

#Store specifc sales graph
def store_wise_sale():
    df = read_excel(data_FILE)
    store = input("Enter store name: ")
    
    # Filter rows where 'StoreLocation' column from dataset matches the given store name.
    result = df[df["StoreLocation"].astype(str) == store]

    # Then group by 'Product' and sum the 'TotalPrice' for each product.
    summary = result.groupby("Product", as_index=False)["TotalPrice"].sum()

    # Plot results
    plt.figure(figsize=(10, 5))
    bars = plt.bar(summary["Product"].astype(str), summary["TotalPrice"])
    plt.xlabel("Product")
    plt.ylabel("Total Price")
    plt.title(f"Sales in Store: {store}")
    plt.xticks(rotation=45)
    plt.bar_label(bars, fmt='%.2f', fontsize=9)
    plt.tight_layout()
    plt.show()

#Salesperson specifc sales graph
def salesperson_wise_sale():
    df = read_excel(data_FILE)
    person = input("Enter salesperson name: ")
    # Filter rows where 'Salesperson' column from dataset matches the given salesperson name.
    result = df[df["Salesperson"].astype(str) == person]

    # Then group by 'Product' and sum the 'TotalPrice' for each product.
    summary = result.groupby("Product", as_index=False)["TotalPrice"].sum()

    # Plot results
    plt.figure(figsize=(10, 5))
    bars = plt.bar(summary["Product"].astype(str), summary["TotalPrice"])
    plt.xlabel("Product")
    plt.ylabel("Total Price")
    plt.title(f"Sales by Salesperson: {person}")
    plt.xticks(rotation=45)
    plt.bar_label(bars, fmt='%.2f', fontsize=9)
    plt.tight_layout()
    plt.show()

#Maximum Product sales in store graph
def max_product_sales_by_store():
    df = read_excel(data_FILE)

    # Group by Store and Product to find total price per product per store
    grouped = df.groupby(["StoreLocation", "Product"], as_index=False)["TotalPrice"].sum()

    # Find max-selling product for each store
    max_products = grouped.loc[grouped.groupby("StoreLocation")["TotalPrice"].idxmax()]

    print("\n=== Max Product Sales by Store ===")
    print(max_products)

    # Plot results
    plt.figure(figsize=(12, 6))
    x = max_products["StoreLocation"].astype(str)
    y = max_products["TotalPrice"]
    bars = plt.bar(x, y)
    plt.xlabel("Store Location")
    plt.ylabel("Total Sales (TotalPrice)")
    plt.title("Maximum Product Sales by Store")
    plt.xticks(rotation=45)
    plt.bar_label(bars, fmt='%.2f', fontsize=9)
    plt.tight_layout()
    plt.show()

#Maximum sales by salesperson graph
def salesperson_max_sales():
    df = read_excel(data_FILE)

    # Group total sales by salesperson
    grouped = df.groupby("Salesperson", as_index=False)["TotalPrice"].sum()

    # Identify max salesperson
    max_index = grouped["TotalPrice"].idxmax()
    max_name = grouped.loc[max_index, "Salesperson"]

    # Plot results
    plt.figure(figsize=(12, 6))
    x = grouped["Salesperson"].astype(str)
    y = grouped["TotalPrice"]
    colors = ["grey"] * len(grouped)
    colors[max_index] = "orange"
    bars = plt.bar(x, y, color=colors)
    plt.title("Total Sales by Salesperson (Max Highlighted)")
    plt.xlabel("Salesperson")
    plt.ylabel("Total Sales")
    plt.xticks(rotation=45)
    plt.bar_label(bars, fmt='%.2f', fontsize=9)
    plt.tight_layout()
    plt.show()

#Returned products for specific store graph
def store_wise_return():
    df = read_excel(data_FILE)
    store = input("Enter store name: ")

    # Filter rows where 'StoreLocation' column from dataset matches the given store name.
    result = df[df["StoreLocation"].str.lower() == store.lower()]

    # Group by product to summarize returns
    grouped = result.groupby("Product", as_index=False)["Returned"].sum()

    # Plot graph
    plt.figure(figsize=(10, 5))
    bars = plt.bar(grouped["Product"], grouped["Returned"])
    plt.title(f"Store-wise Return for {store}")
    plt.xlabel("Product")
    plt.ylabel("Total Returned")
    plt.xticks(rotation=45)
    plt.bar_label(bars, fmt='%.2f', fontsize=9)
    plt.tight_layout()
    plt.show()
```
